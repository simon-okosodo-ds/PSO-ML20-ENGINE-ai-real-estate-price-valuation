# PHASE 3: Data Cleaning: Drop Identifiers & Duplicates
**What was done:** We manually inspected columns to identify and remove non-predictive identifiers, eliminated duplicate records, and reset the dataframe index.

**Method used:** df.drop() with errors='ignore', df.drop_duplicates(), and df.reset_index().
Strategy employed: "Noise Elimination & Data Integrity" Strategy.

**Reason for method:** Using errors='ignore' makes the script robust against re-runs, while drop_duplicates() ensures the model doesn't over-train on identical data points (preventing bias).

**Reason for strategy:** Identifiers like Serial_Number or Full_Name are unique to individuals, not house quality. If left in, a model might "memorize" specific names (overfitting) rather than learning the actual physics of house pricing. Removing duplicates ensures each data point provides a unique signal.

```python
# 1. Define non-predictive columns
cols_to_remove = ['Serial_Number', 'Full_Name', 'Description', 'id', 'PropertyID']

# 2. Record initial counts for summary
initial_rows, initial_cols = df.shape

# 3. Execute drops, duplicate removal, and index reset
df = df.drop(columns=cols_to_remove, errors='ignore')
df = df.drop_duplicates()
df = df.reset_index(drop=True)

# 4. Reporting
final_rows, final_cols = df.shape
print(f"✅ Cleanup Complete: Dropped {initial_cols - final_cols} columns.")
print(f"✅ Deduplication: Removed {initial_rows - final_rows} duplicate rows.")
print(f"📊 Final structure: {final_rows} rows | {final_cols} features.")
```

# PHASE 4. Data Cleaning: Row Integrity Filter.

**What was done:** We evaluated the percentage of missing data in each row and removed those with more than 50% missing values.

**Method used:** df.isnull().mean(axis=1) to calculate row-wise gaps and .loc[] for filtering.

**Strategy employed:** "Integrity Threshold" Filtering.

**Reason for method:** Rows missing more than half their data are essentially "empty shells." Imputing values for 50%+ of a row would introduce artificial "hallucinated" data that confuses the model.

**Reason for strategy:** Removing these rows ensures the model trains only on real, substantial patterns. Resetting the index ensures no "holes" remain in the dataframe's address system, preventing future alignment errors.

```python
# Calculate the percentage of missing values per row
row_missing_percent = df.isnull().mean(axis=1) * 100

# Identify counts before filter for reporting
initial_rows = len(df)

# Keep only rows that have at least 50% of their data present
df = df.loc[row_missing_percent < 50, :]
df = df.reset_index(drop=True)

# Simple summary report
dropped_rows = initial_rows - len(df)
print(f"✅ Row Filter Complete: Removed {dropped_rows} low-quality rows. {len(df)} rows remaining.")
```

# PHASE 5. Data Cleaning: Column Integrity Filter.
**What was done:** We calculated the percentage of missing values for every column and removed those missing more than 50% of their data.

**Method used:** df.isnull().mean() to find column-wise gap ratios and .loc[:, mask] to filter the DataFrame.

**Strategy employed** "Informational Value" Thresholding.

**Reason for method:** mean() provides a clear 0 to 1 ratio, which is easily converted to a percentage. Filtering using a boolean mask is the most memory-efficient way to drop multiple columns at once.

**Reason for strategy:** Features missing more than half their values are considered "sparse." Attempting to fill (impute) such a large gap usually introduces bias or "noise," as the model would be learning from mostly estimated data rather than real-world signals.

```python
# 1. Calculate missing data percentage per column
missing_percent = df.isnull().mean() * 100

# 2. Log count before dropping for the summary
cols_before = df.shape[1]

# 3. Keep only columns with less than 50% missing data
df = df.loc[:, missing_percent < 50]

# 4. Reporting (No noisy lists)
cols_dropped = cols_before - df.shape[1]
print(f"✅ Column Filter Complete: Dropped {cols_dropped} sparse features. {df.shape[1]} reliable features remaining.")
```

# PHASE 8: Automated Feature Synthesis (Ratio Engineering)
**What was done:** We engineered a diagnostic function to automatically discover hidden mathematical relationships between numeric variables. The system systematically pairs every numeric feature (e.g., sqft_living, bedrooms, bathrooms) to calculate their ratios and measures their individual correlation strength against the target price.

**Strategy employed:** "Automated Heuristic Proportionality" Strategy. By testing every possible combination of numerator and denominator, we move beyond human intuition to identify features that represent "Value per Unit" (e.g., living space per room or lot size relative to floor count).

**Reason behind strategy choice:** In high-stakes markets like real estate, absolute numbers don’t tell the whole story. A large house on a tiny lot has a completely different value profile than a medium house on a massive lot. By automating the search for these proportions, we ensure the model captures the "Density" and "Balance" of property attributes, which are often more predictive than raw counts alone.

**Our Result:** We successfully isolated the Top 10 Ratio Pairs with the highest statistical correlation to price. These "Synthetic Signals" have been ranked and shortlisted, providing the model with a "shortcut" to understanding property proportionality before we proceed to the final training phases.

```python
def generate_best_ratio_pairs(df, target, top_k=10):

    df = df.copy()

    # Keep only numeric columns
    num_cols = df.select_dtypes(include=np.number).columns.tolist()

    # Remove target from candidates
    num_cols = [col for col in num_cols if col != target]

    results = []

    for i in range(len(num_cols)):
        for j in range(len(num_cols)):
            if i != j:
                col1, col2 = num_cols[i], num_cols[j]

                try:
                    new_feature = df[col1] / (df[col2] + 1)

                    corr = new_feature.corr(df[target])

                    if not np.isnan(corr):
                        results.append((col1, col2, abs(corr)))
                except:
                    continue

    # Convert to DataFrame
    scores_df = pd.DataFrame(results, columns=['Numerator', 'Denominator', 'Score'])

    # Sort by best correlation
    scores_df = scores_df.sort_values(by='Score', ascending=False)

    # Get top pairs
    best_pairs = list(zip(scores_df['Numerator'], scores_df['Denominator']))[:top_k]

    return best_pairs, scores_df

TARGET_COL = 'SalePrice'   # 🔁 change if needed

best_ratio_pairs, ratio_scores = generate_best_ratio_pairs(
    df,
    target=TARGET_COL,
    top_k=10
)

print("✅ Best Ratio Pairs:")
print(best_ratio_pairs)
```

# PHASE 9: Automated Feature Synthesis (Non-Linear Polynomials)
**What was done:** We developed a diagnostic script to detect non-linear relationships by calculating the squares of numeric features and measuring their correlation with the target price. The function isolates features where a "curved" relationship (e.g., exponential value growth relative to size) offers a stronger predictive signal than the original linear data.

**Strategy employed:** "Quadratic Signal Discovery" Strategy. This method targets variables that exhibit "diminishing returns" or "accelerating value" patterns—complex behaviors that standard linear models often fail to capture without manual intervention.

**Reason behind strategy choice:** Real-world market variables rarely follow perfectly straight lines. For instance, the impact of house_age or sqft_living on price often intensifies at the extremes (brand new homes vs. historic estates). By testing squared versions of these columns, we capture the mathematical "bends" in the data, providing the model with a higher-resolution understanding of value acceleration.

**Our Result:** The function successfully identified and isolated the Top 5 Polynomial candidates. This ensures we only introduce non-linear complexity where it is statistically proven to add value, keeping our feature set concentrated and high-quality.

```python
def generate_best_poly_cols(df, target, top_k=10):

    df = df.copy()

    # Keep numeric columns only
    num_cols = df.select_dtypes(include=np.number).columns.tolist()

    # Remove target
    num_cols = [col for col in num_cols if col != target]

    results = []

    for col in num_cols:
        try:
            squared = df[col] ** 2

            corr = squared.corr(df[target])

            if not np.isnan(corr):
                results.append((col, abs(corr)))
        except:
            continue

    # Create dataframe
    scores_df = pd.DataFrame(results, columns=['Column', 'Score'])

    # Sort by best correlation
    scores_df = scores_df.sort_values(by='Score', ascending=False)

    # Get top columns
    best_cols = scores_df['Column'].head(top_k).tolist()

    return best_cols, scores_df

TARGET_COL = 'SalePrice'  # 🔁 change if needed

best_poly_cols, poly_scores = generate_best_poly_cols(
    df,
    target=TARGET_COL,
    top_k=5
)

print("✅ Best Polynomial Columns:")
print(best_poly_cols)
```

# PHASE 10: Strategic Feature Deployment (Manual Engineering)
**What was done:** We integrated the top-performing automated discoveries (ratios and polynomials) with custom domain logic. This function physically transformed the dataset by creating "Value per Unit" features, calculating property age from the current year, squaring high-impact variables, and decomposing raw date strings into granular time signals (Year, Month, Weekday, and Weekend status).

**Strategy employed:** "Multi-Dimensional Feature Synthesis" Strategy. This consolidates automated statistical findings with temporal and structural engineering to provide the model with a 360-degree view of the data.

**Reason behind strategy choice:** Raw data is rarely in its most predictive form. For example, a house's price is often dictated by its age (how many years since it was built) and timing (seasonality in the housing market), rather than just the raw year of construction. By manually anchoring the best ratios and date parts into the dataset, we "pre-digest" the complex relationships for the model, making the learning process faster and more accurate.

**Our Result:** The dataset has been successfully enriched. We transitioned from basic raw columns to a sophisticated set of predictors that account for proportionality, time-based trends, and property depreciation (age). The data is now fully prepared for the final Pipeline assembly.

```python
from datetime import datetime
# ==========================================
# 1. USER CONFIGURATION (ADJUST HERE), Never add the Target Column
# ==========================================
MY_RATIO_PAIRS =  best_ratio_pairs #[('price', 'bedrooms'), ('sqft_living', 'bedrooms')]
MY_POLY_COLS    = best_poly_cols #['sqft_living']
MY_DATE_COLS    = ['DocumentDate']
MY_AGE_MAP      = {'house_age': 'YrBuilt'}  # {'New_Name': 'Old_Year_Column'}

# ==========================================
# 2. THE FUNCTION (RUN ONCE)
# ==========================================
def create_manual_features(df, ratio_pairs=[], poly_cols=[], date_cols=[], age_map={}):
    df = df.copy()

    # Ratios: Creates "Value per Unit" (e.g., Price per Room)
    for a, b in ratio_pairs:
        if a in df.columns and b in df.columns:
            df[f'{a}_per_{b}'] = df[a] / (df[b] + 1) # +1 prevents division by zero

    # Age: Turns a Year (1995) into an Age (31 years old)
    current_year = datetime.now().year
    for new_name, col in age_map.items():
        if col in df.columns:
            df[new_name] = current_year - df[col]

    # Polynomials: Helps Linear models see curved patterns
    for col in poly_cols:
        if col in df.columns:
            df[f'{col}_squared'] = df[col] ** 2

    # Dates: Breaks a single date into Month, Year, Weekend, etc.
    for col in date_cols:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')
            df[f'{col}_year'] = df[col].dt.year
            df[f'{col}_month'] = df[col].dt.month
            df[f'{col}_weekday'] = df[col].dt.weekday
            df[f'{col}_is_weekend'] = df[col].dt.weekday.isin([5,6]).astype(int)

    return df

# ==========================================
# 3. EXECUTION
# ==========================================
df = create_manual_features(
    df,
    ratio_pairs=MY_RATIO_PAIRS,
    poly_cols=MY_POLY_COLS,
    date_cols=MY_DATE_COLS,
    age_map=MY_AGE_MAP
)

print(df.head())
```

# PHASE 12: Final Feature Selection (The "Strict Judge")

**What was done:** We implemented a high-stakes "Feature Audit" to filter the massive interaction matrix created in the previous phases. Utilizing a Regularized Random Forest, the system performed a survival test on every column, ranking them by their actual predictive contribution. We implemented Leakage Shielding to drop non-predictive variables and sanitized the remaining features (fixing infinity/nulls) to ensure the "Judge" had a clean mathematical environment for evaluation.

**Strategy employed:** "Survival of the Fittest" Strategy. Instead of assuming all engineered interaction terms were useful, we enforced a strict selection process. **Reason behind strategy choice:** After Phase 11, the dataset likely grew to hundreds of "synthetic" interaction terms. Many of these are redundant or specific to single houses. # PHASE 13: Universal Pipeline Engine Assembly
**What was done:** We engineered a sophisticated, automated Pipeline factory that dynamically adapts to the data. The function performs a statistical audit of the "Gold" features, automatically assigning PowerTransformers to skewed inputs and Target Encoding to high-cardinality categories. It also wraps the final model in a TransformedTargetRegressor to handle non-linear target distributions (like house prices or store sales) through logarithmic scaling.

**Strategy employed:** "Adaptive Transformation Architecture" Strategy. This moves away from "static" coding by creating a system that "reads" the data's statistical properties and assigns the appropriate preprocessing tools (e.g., RobustScaler for outliers or Yeo-Johnson for skewed distributions) on the fly.

**Reason behind strategy choice:** Real-world data is notoriously messy—targets are often skewed, and features like sqft_living or CompetitionDistance have massive outliers. This "Universal" approach ensures each column receives the specific mathematical treatment it needs to perform best in a Gradient Boosting model. By automating this, we eliminate human error and ensure the model treats a 10% error on a small value with the same weight as a 10% error on a large value.

**Our Result:** The system generated a Comprehensive Strategy Report, confirming it has successfully grouped our "Winner" features into optimal processing buckets. We now have a single, leak-proof pipeline object that handles cleaning, scaling, encoding, and modeling in one unified, production-ready workflow.

```python
from sklearn.preprocessing import StandardScaler, RobustScaler, PowerTransformer, OneHotEncoder, OrdinalEncoder
!pip install category_encoders
from category_encoders import TargetEncoder
from xgboost import XGBRegressor

def build_universal_pipeline(df_final, target_col, ordinal_mapping=None):
    """
    UNIVERSAL DATA PIPELINE ENGINE (Optimized for df_final)
    """
    # Initialize ordinal mapping if not provided
    ord_map = ordinal_mapping if ordinal_mapping else {}

    # Separate Features and Target using df_final
    X = df_final.drop(columns=[target_col])
    y = df_final[target_col]

    # Initial Type Separation
    num_df = X.select_dtypes(include='number')
    cat_df = X.select_dtypes(exclude='number')
    ord_cols = list(ord_map.keys())

    # 1. REMOVE ORDINALS FROM AUTO-DETECTION
    rem_cat = cat_df.drop(columns=[c for c in ord_cols if c in cat_df.columns]) if ord_cols else cat_df
    rem_num = num_df.drop(columns=[c for c in ord_cols if c in num_df.columns]) if ord_cols else num_df

    # 2. AUTO-GROUP NUMERICALS
    power_robust_cols = rem_num.columns[rem_num.skew().abs() > 2].tolist()
    power_scale_cols = rem_num.columns[(rem_num.skew().abs() > 0.75) & (rem_num.skew().abs() <= 2)].tolist()
    standard_cols = rem_num.columns[rem_num.skew().abs() <= 0.75].tolist()

    # 3. AUTO-GROUP CATEGORICALS
    high_nom = rem_cat.columns[rem_cat.nunique() > 30].tolist()
    low_nom = rem_cat.columns[(rem_cat.nunique() <= 30) & (rem_cat.nunique() > 2)].tolist()
    bin_cols = rem_cat.columns[rem_cat.nunique() == 2].tolist()

    # 4. TARGET ANALYSIS
    target_skew = y.skew()
    if abs(target_skew) > 2:
        t_action, t_trans = "PowerTransformer (Yeo-Johnson)", PowerTransformer(method='yeo-johnson')
    elif abs(target_skew) > 0.5:
        t_action, t_trans = "Log Transformation (log1p)", "log1p"
    else:
        t_action, t_trans = "Standard (No Transformation)", None

    # --- START STRATEGY REPORT ---
    print("\n" + "="*80)
    print(f"📊 UNIVERSAL PIPELINE STRATEGY REPORT")
    print("="*80)
    print(f"1: INITIAL DATA TYPE GROUPING")
    print(f"   - Numerical features : {len(rem_num.columns)}")
    print(f"   - Binary categories  : {len(bin_cols)}")
    print(f"   - Nominal categories : {len(high_nom + low_nom)}")
    print(f"   - Ordinal category   : {len(ord_cols)}")
    print("-" * 80)

    print(f"2. TARGET [{target_col}]:")
    print(f"   - SKEW: {target_skew:.2f}")
    print(f"   - ACTION: {t_action}")
    print("-" * 40)

    # --- ASSEMBLE TRANSFORMERS ---
    transformers = []
    if power_robust_cols:
        transformers.append(('ext_num', Pipeline([('im', SimpleImputer(strategy='median')), ('pt', PowerTransformer()), ('rs', RobustScaler())]), power_robust_cols))
    if power_scale_cols:
        transformers.append(('mod_num', Pipeline([('im', SimpleImputer(strategy='median')), ('pt', PowerTransformer()), ('ss', StandardScaler())]), power_scale_cols))
    if standard_cols:
        transformers.append(('norm_num', Pipeline([('im', SimpleImputer(strategy='mean')), ('ss', StandardScaler())]), standard_cols))
    if high_nom:
        transformers.append(('hi_nom', Pipeline([('im', SimpleImputer(strategy='constant', fill_value='Unknown')), ('te', TargetEncoder())]), high_nom))
    if low_nom:
        transformers.append(('lo_nom', Pipeline([('im', SimpleImputer(strategy='most_frequent')), ('ohe', OneHotEncoder(handle_unknown='ignore'))]), low_nom))
    if bin_cols:
        transformers.append(('bin', Pipeline([('im', SimpleImputer(strategy='most_frequent')), ('oe', OrdinalEncoder())]), bin_cols))
    if ord_cols:
        cats = [ord_map[c] for c in ord_cols]
        transformers.append(('manual_ord', Pipeline([('im', SimpleImputer(strategy='most_frequent')), ('oe', OrdinalEncoder(categories=cats))]), ord_cols))

    # --- FINAL PIPELINE GROUPS ---
    print("🚀3: FINAL PIPELINE GROUPS")
    active_groups = {
        'PowerTransformer (Skewed/Outliers)': power_robust_cols + power_scale_cols,
        'StandardScaler (Normal)': standard_cols,
        'TargetEncoder (High-Cat)': high_nom,
        'OneHotEncoder (Low-Cat)': low_nom,
        'BinaryMapping': bin_cols,
        'OrdinalEncoder (Manual Mapping)': ord_cols
    }

    for group, cols in active_groups.items():
        if cols:
            print(f"[{group}]: {cols}")
        else:
            print(f"[{group}]: NOT AVAILABLE IN THIS DATASET")

    if not ord_cols:
        print("\n⚠️ NOTE: No Ordinal features were mapped. Ensure you haven't missed ranked features.")

    print("="*80)

    preprocessor = ColumnTransformer(transformers=transformers, remainder='drop')
    model = TransformedTargetRegressor(regressor=XGBRegressor(n_estimators=100), transformer=t_trans)

    return Pipeline([('preprocessor', preprocessor), ('model', model)])

my_ord_map = {} # Edit this as needed per problem

# Run the build
pipeline = build_universal_pipeline(df_final, target_col='SalePrice', ordinal_mapping=my_ord_map)
```

# PHASE 16: Adaptive Data Hardening & Structural Alignment
**What was done:** We implemented an intelligent "Data Hardening" utility that adapts its logic based on the project’s mathematical nature. For Regression tasks, the system enforced zero-tolerance boundaries (protecting against Log-transform errors) and stabilized variance through a 1% tail-clip. For all tasks, the system performed an automated High-Cardinality Audit, dropping any features that behaved like unique identifiers rather than predictive patterns. The phase concluded with a structural index realignment to ensure 100% positional synchronization.

**Strategy employed:** "Context-Aware Sanitization" Strategy. This involves a secondary, automated check for Memorization Bias. By auditing the uniqueness ratio of every feature, the system neutralizes "hidden identifiers" that escaped the initial manual cleanup.

**Reason behind strategy choice:** Manual cleanup often misses features like specific address strings or transaction hashes that are mathematically identical to IDs. Enforcing a uniqueness threshold ensure the model learns generalizable market trends rather than specific row-level data. Resetting the index "heals" the dataset's skeleton after multiple rounds of row removal.

```python
# ⌨️ 🛑 USER INPUT: SET SETTINGS
TASK = 'reg' # 'reg' or 'clf'
TARGET_COL = 'SalePrice'

def apply_data_hardening(df, target, task='reg'):
    df_clean = df.copy()
    initial_rows = len(df_clean)

    print(f"🛠️ Hardening Protocol: {task.upper()} Mode Active")
    print("-" * 55)

    # 1. MATHEMATICAL BOUNDARY ENFORCEMENT (Regression Only)
    if task == 'reg' and target in df_clean.columns:
        # Remove impossible values for log-math
        df_clean = df_clean[df_clean[target] > 0]
        # Target Stabilization (Final 1% Tail-Censorship)
        upper_limit = df_clean[target].quantile(0.99)
        df_clean = df_clean[df_clean[target] < upper_limit]
        print(f"✅ Regression: Enforced >0 boundaries and 99% tail-clip.")

    # 2. AGNOSTIC CARDINALITY AUDIT (The "Safety Net")
    # Finds columns that escaped Phase 3 but still act like unique IDs
    cols_to_check = [c for c in df_clean.columns if c != target]
    for col in cols_to_check:
        uniqueness_ratio = df_clean[col].nunique() / len(df_clean)
        # If more than 90% of rows are unique, it's a 'Memorization Risk'
        if uniqueness_ratio > 0.9:
            df_clean = df_clean.drop(columns=[col])
            print(f"🗑️ Dropped '{col}': High Cardinality detected ({uniqueness_ratio:.1%} unique).")

    # 3. PHYSICAL INDEX ALIGNMENT (Essential for safe training)
    df_clean = df_clean.reset_index(drop=True)

    print(f"✅ Index Reset: {len(df_clean)} continuous rows synchronized.")
    print(f"📊 Total Rows Removed in this phase: {initial_rows - len(df_clean)}")
    print("-" * 55)

    return df_clean

# 🏁 EXECUTE
df_final = apply_data_hardening(df_final, TARGET_COL, task=TASK)
```

# PHASE 17: Master Pipeline Architecture (The Adaptive Engine)
**What was done:** We engineered a modular "Master Pipeline" factory that acts as the project's central intelligence. The function performs four critical tasks: auto-groups features based on statistical skewness, implements high-cardinality Target Encoding to neutralize "categorical explosion," forces consistent Pandas output for total traceability, and utilizes a Task-Aware Wrapper. For regression, it dynamically applies target log-transformations to stabilize high-variance predictions.

**Strategy employed:** "Statistical-Aware Pipeline Injection" Strategy. Instead of static steps, the pipeline self-configures its mathematical transformations (Yeo-Johnson vs. Standard scaling) based on the physical properties of the features. It uses "Lazy Grouping" to detect column types and skewness at runtime, ensuring the model always receives "optimal signal."

**Reason behind strategy choice:** Real-world data is notoriously messy—house prices are skewed, Titanic features are mixed, and Fraud datasets are massive. By using Yeo-Johnson PowerTransformers for skewed columns and TargetEncoders for hundreds of unique categories (like streets or IDs), we solve the problems of non-normal distributions and memory bloat. The TransformedTargetRegressor is the "secret sauce" for regression, ensuring the model works on a normalized scale without losing original dollar value.

**Our Result:** We have created a "Universal Engine" that is both robust and flexible. It is designed to work as a template for any machine learning task. It successfully handles the extreme skewness observed in our data audits and is now ready to be plugged into a competitive audition loop.

```python

import sklearn
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PowerTransformer, OneHotEncoder
from category_encoders import TargetEncoder

def build_universal_pipeline(df_final, target_col, model_obj, task='reg'):
    """
    UNIVERSAL DATA PIPELINE ENGINE (Adaptive for Regression & Classification)
    - df_final: The input dataframe
    - target_col: The target variable name
    - model_obj: The ML model (XGBoost, CatBoost, etc.)
    - task: 'reg' for Regression, 'clf' for Classification
    """
    # Force Pandas output for total traceability
    sklearn.set_config(transform_output="pandas")

    # Prepare features
    X = df_final.drop(columns=[target_col], errors='ignore')
    y = df_final[target_col]

    # --- STEP 1: AUTOMATED FEATURE GROUPING ---
    num_df = X.select_dtypes(include='number')
    cat_df = X.select_dtypes(exclude='number')

    # Skew detection for Numerics (Threshold: 0.75)
    power_cols = num_df.columns[num_df.skew().abs() > 0.75].tolist()
    standard_cols = num_df.columns[num_df.skew().abs() <= 0.75].tolist()

    # Cardinality Audit (High vs Low)
    high_nom = cat_df.columns[cat_df.nunique() > 30].tolist()
    low_nom = cat_df.columns[cat_df.nunique() <= 30].tolist()

    # --- STEP 2: BUILD TRANSFORMERS (Error-Proof System) ---
    transformers = []

    if power_cols:
        transformers.append(('pow', Pipeline([
            ('im', SimpleImputer(strategy='median')),
            ('pt', PowerTransformer(method='yeo-johnson')),
            ('ss', StandardScaler())
        ]), power_cols))

    if standard_cols:
        transformers.append(('std', Pipeline([
            ('im', SimpleImputer(strategy='mean')),
            ('ss', StandardScaler())
        ]), standard_cols))

    if high_nom:
        transformers.append(('hi', Pipeline([
            ('im', SimpleImputer(strategy='constant', fill_value='Missing')),
            ('te', TargetEncoder())
        ]), high_nom))

    if low_nom:
        transformers.append(('lo', Pipeline([
            ('im', SimpleImputer(strategy='most_frequent')),
            ('ohe', OneHotEncoder(handle_unknown='ignore', sparse_output=False, drop='if_binary'))
        ]), low_nom))

    preprocessor = ColumnTransformer(transformers=transformers, remainder='passthrough')

    # --- STEP 3: DYNAMIC TARGET WRAPPING ---
    if task == 'reg':
        # REGRESSION: Check if Target Skew needs Log-transform
        target_skew = y.skew()
        if abs(target_skew) > 0.75:
            final_model = TransformedTargetRegressor(
                regressor=model_obj,
                func=np.log1p,
                inverse_func=np.expm1
            )
        else:
            final_model = model_obj
    else:
        # CLASSIFICATION: Use model as-is
        final_model = model_obj

    # --- STEP 4: FINAL PIPELINE ASSEMBLY ---
    return Pipeline([
        ('preprocessor', preprocessor),
        ('model', final_model)
    ])
```
